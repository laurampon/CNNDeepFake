```python
source_df = pd.DataFrame(
    fake_sources.items(),
    columns=["generator", "count"]
).sort_values("count")

fig, ax = plt.subplots(figsize=(8, 9))

ax.barh(
    source_df["generator"],
    source_df["count"],
    color="#D47A52",
    height=0.7
)

for i, value in enumerate(source_df["count"]):
    ax.text(value + 0.3, i, str(value), va="center", fontsize=10)

ax.set_xlabel("Número de imagens", fontsize=10)
ax.set_ylabel("Gerador", fontsize=10)
ax.set_title("Imagens sintéticas por tipo de gerador", weight="bold", fontsize=12)

ax.set_xlim(0, 75)
ax.grid(axis="x")

plt.tight_layout()
plt.savefig("./results/images/fig_generators_distribution.png")
plt.show()
```
